## Bivariate Analysis / Multivariate Analysis

```python
# 1. Does the cost of goods sold affect the ratings that the customers provide
sns.scatterplot(x=data['cogs'], y=data['Rating'])
```

```python
# 2. Does gross income affect the ratings that the customers provide
data.columns
```

```python
sns.scatterplot(x=data['gross income'], y=data['Rating'])
```

```python
# 3. Find the most profitable branch as per gross income 
plt.figure(figsize=(8,4))
sns.barplot(x=data['Branch'], y=data['gross income'], ci=None, hue=data['City'])
```

```python
# 4. Is there any relationship between gender and gross income 
sns.barplot(x=data['Gender'], y=data['gross income'], hue=data['Customer type'], ci=None)
```

```python
# 5. Find the product line that generates the most income 
data.columns
```

```python
plt.figure(figsize=(10,3))
plt.xticks(rotation=65)
sns.barplot(x=data['Product line'], y=data['gross income'], ci=None, hue=data['Product line'])
```

```python
# 6. Find the highest unit price in the product line 
plt.figure(figsize=(10,3))
plt.xticks(rotation=65)
sns.barplot(x=data['Product line'], y=data['Unit price'], ci=None, hue=data['Product line'])
```

```python
data.groupby('Product line')['Unit price'].max().sort_values(ascending=False).head(3)
```

```python
# 7. Find different payment methods used by customers citywise
data.columns
```

```python
data.groupby('City')['Payment'].value_counts()
```

```python
paymnt_by_city = pd.crosstab(data['Payment'], data['City'])
paymnt_by_city
```

```python
plt.figure(figsize=(8,4))
sns.heatmap(paymnt_by_city)
```

```python
# 8. Which product line is purchased in the highest quantity 
highest_quant = data.groupby('Product line')['Quantity'].sum()
highest_quant
```

```python
plt.figure(figsize=(12,4))
sns.barplot(highest_quant)
```

```python
# 9. Display daily sales by day of the week 
data.columns
```

```python
day_map = {0:'Mon', 1:'Tue', 2:'Wed', 3:'Thu', 4:'Fri', 5:'Sat', 6:'Sun'}
```

```python
data['Day of Week'] = data['Date'].dt.dayofweek.map(day_map)
```

```python
data.head(3)
```

```python
data['Day of Week'].value_counts()
```

```python
sns.countplot(x=data['Day of Week'])
```

```python
# 10. What will be the highest months for sales 
data['Date'].dt.month
```

```python
month_map = {1:'January', 2:'February', 3:'March'}
```
